```python
import curses
import psutil
import time

def draw_monitor(stdscr):
    curses.curs_set(0)

    curses.start_color()
    curses.init_pair(1, curses.COLOR_GREEN, curses.COLOR_BLACK)
    curses.init_pair(2, curses.COLOR_YELLOW, curses.COLOR_BLACK)
    curses.init_pair(3, curses.COLOR_RED, curses.COLOR_BLACK)

    while True:
        stdscr.clear()

        # System stats
        cpu = psutil.cpu_percent()
        ram = psutil.virtual_memory().percent
        net = psutil.net_io_counters()

        # Choose color based on usage
        def get_color(value):
            if value < 50:
                return curses.color_pair(1)
            elif value < 80:
                return curses.color_pair(2)
            else:
                return curses.color_pair(3)

        # Draw progress bar
        def draw_bar(y, label, value):
            bar_length = 40
            filled = int(bar_length * value / 100)

            bar = "█" * filled + "-" * (bar_length - filled)

            stdscr.addstr(
                y,
                2,
                f"{label}: [{bar}] {value:.1f}%",
                get_color(value)
            )

        # Title
        stdscr.addstr(
            1,
            2,
            "Live System Monitor",
            curses.color_pair(1)
        )

        stdscr.addstr(
            2,
            2,
            "_" * 55,
            curses.color_pair(1)
        )

        # CPU and RAM bars
        draw_bar(4, "CPU", cpu)
        draw_bar(6, "RAM", ram)

        # Network usage
        stdscr.addstr(
            8,
            2,
            f"NET Sent: {net.bytes_sent / 1024 / 1024:.2f} MB   "
            f"Recv: {net.bytes_recv / 1024 / 1024:.2f} MB",
            curses.color_pair(1)
        )

        stdscr.addstr(
            10,
            2,
            "Press CTRL+C to exit",
            curses.color_pair(2)
        )

        stdscr.refresh()
        time.sleep(1)

# Start app
curses.wrapper(draw_monitor)
```

```python
import psutil
import time
from IPython.display import clear_output

while True:
    clear_output(wait=True)

    cpu = psutil.cpu_percent()
    ram = psutil.virtual_memory().percent
    net = psutil.net_io_counters()

    print("===== Live System Monitor =====")
    print(f"CPU Usage : {cpu}%")
    print(f"RAM Usage : {ram}%")
    print(f"Sent Data : {net.bytes_sent / 1024 / 1024:.2f} MB")
    print(f"Recv Data : {net.bytes_recv / 1024 / 1024:.2f} MB")

    time.sleep(1)
```
